# Finite Difference Approximations

We approximate the first and second derivatives of $f(x) = \sin(x)$
using the standard second-order centered differences
$$
D_1 f(x) = \frac{f(x+h)-f(x-h)}{2h}, \qquad
D_2 f(x) = \frac{f(x+h)-2f(x)+f(x-h)}{h^2}.
$$
We measure the maximum error on $n$ equally spaced points in $[0, 2\pi)$,
with $h = 2\pi/n$. We double $n$ each time (which halves $h$) and so the
observed order of convergence is $p = \log_2(E_{\mathrm{prev}}/E)$.

```python
import numpy as np
import matplotlib.pyplot as plt


def centered_differences(x, h):
    fm, f, fp = np.sin(x - h), np.sin(x), np.sin(x + h)
    first = (fp - fm) / (2 * h)
    second = (fp - 2 * f + fm) / h**2
    return first, second
```

```python
header = f"{'points':>6} {'h':>8} | {'1st deriv error':>12} {'p':>2} | {'2nd deriv error':>12} {'p':>2}"
print(header)
print("-" * len(header))

previous = None
for n in (8, 16, 32, 64, 128):
    h = 2 * np.pi / n
    x = h * np.arange(n)
    first, second = centered_differences(x, h)
    errors = [np.max(np.abs(np.cos(x) - first)), np.max(np.abs(-np.sin(x) - second))]
    row = f"{n:6d} {h:8.4f}"
    for k, error in enumerate(errors):
        rate = f"{np.log2(previous[k] / error):5.2f}" if previous is not None else " " * 5
        row += f" | {error:12.3e} {rate}"
    print(row)
    previous = errors
```

## First and second derivatives on a 16-point grid

```python
n = 16
h = 2 * np.pi / n
x = h * np.arange(n)
first, second = centered_differences(x, h)
t = np.linspace(0, 2 * np.pi, 201)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(9, 3), sharey=True, layout="constrained")
ax1.plot(t, np.cos(t), "k-", label="Exact")
ax1.plot(x, first, "o", label="Centered difference")
ax1.set_title("First derivative")
ax2.plot(t, -np.sin(t), "k-", label="Exact")
ax2.plot(x, second, "o", label="Centered difference")
ax2.set_title("Second derivative")
for ax in (ax1, ax2):
    ax.set_xlabel("$x$")
    ax.legend()
plt.show()
```

## A plot

The function $e^{-x^2}$ on $[-2, 2]$: a bell-shaped curve with its maximum value of 1 at $x = 0$, decreasing symmetrically toward 0 on both sides.

```python
x = np.linspace(-2, 2, 101)
plt.plot(x, np.exp(-x**2))
plt.xlabel("$x$")
plt.ylabel("$e^{-x^2}$")
plt.show()
```

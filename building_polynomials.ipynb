# Building Polynomials

Given distinct nodes $x_0, \ldots, x_n$, the Lagrange polynomials are
$$
\ell_j(x) = \prod_{\substack{k=0\\k\neq j}}^n \frac{x - x_k}{x_j - x_k},
$$
and they satisfy $\ell_j(x_i) = \delta_{ij}$.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def lagrange(x, nodes, j):
    """The j-th Lagrange polynomial for the given nodes, evaluated at x."""
    ell = np.ones_like(x)
    for k, xk in enumerate(nodes):
        if k != j:
            ell *= (x - xk) / (nodes[j] - xk)
    return ell
```

## The Lagrange basis

We take $n = 5$, with $n + 1 = 6$ equidistant nodes on $[-1, 1]$.
Each $\ell_j$ is equal to 1 at $x_j$ and 0 at the other nodes.

```python
n = 5
nodes = np.linspace(-1, 1, n + 1)
x = np.linspace(-1, 1, 401)

for j in range(n + 1):
    plt.plot(x, lagrange(x, nodes, j), label=rf"$\ell_{j}$")
plt.plot(nodes, np.zeros_like(nodes), "ko")
plt.plot(nodes, np.ones_like(nodes), "ko")

plt.xlabel("$x$")
plt.title("Lagrange basis")
plt.legend(loc="upper left", bbox_to_anchor=(1, 1))
plt.show()
```

## Interpolation

Let $f(x) = e^x \sin(3x)$, and $y_j = f(x_j)$.
The interpolating polynomial is
$$
p(x) = \sum_{j=0}^n y_j \ell_j(x).
$$

```python
def f(x):
    return np.exp(x) * np.sin(3 * x)


y = f(nodes)


def p_lagrange(x):
    return sum(y[j] * lagrange(x, nodes, j) for j in range(n + 1))
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5))

ax1.plot(x, f(x), "k", linewidth=2, label="$f$")
ax1.plot(x, p_lagrange(x), "--", label="$p$")
ax1.plot(nodes, y, "ko")
ax1.set_xlabel("$x$")
ax1.set_title(r"$f(x) = e^x \sin(3x)$ and its interpolant")
ax1.legend()

ax2.plot(x, f(x) - p_lagrange(x))
ax2.plot(nodes, np.zeros_like(nodes), "ko")
ax2.set_xlabel("$x$")
ax2.set_title("Error $f - p$")

plt.show()
```

## The Vandermonde matrix

Writing $p(x) = \sum_{j=0}^n a_j x^j$ in the monomial basis, the conditions $p(x_i) = y_i$ give the linear system $V a = y$, where $V_{ij} = x_i^j$.

```python
V = np.zeros((n + 1, n + 1))
for i in range(n + 1):
    for j in range(n + 1):
        V[i, j] = nodes[i] ** j
print("V =")
print(V)
```

```python
a = np.linalg.solve(V, y)
print("a =", a)


def p_monomial(x):
    return sum(a[j] * x**j for j in range(n + 1))
```

By uniqueness of the interpolating polynomial, both approaches give the same polynomial (up to rounding errors).

```python
print("max |p_lagrange - p_monomial| =", np.max(np.abs(p_lagrange(x) - p_monomial(x))))

plt.plot(x, f(x), "k", linewidth=2, label="$f$")
plt.plot(x, p_lagrange(x), label="Lagrange")
plt.plot(x, p_monomial(x), "--", label="Vandermonde")
plt.plot(nodes, y, "ko")

plt.xlabel("$x$")
plt.legend()
plt.show()
```

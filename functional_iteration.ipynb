# Functional Iteration

Example 2.18: the functional iterates $x_{n+1} = g(x_n)$ of $g(x) = \cos(x)$, starting from $x_0 = 3\pi$. The iterates are marked on the graph of $y = g(x)$, i.e. at the points $(x_n, g(x_n)) = (x_n, x_{n+1})$. The iterates converge to $r \approx 0.739085$, where the graph of $y = \cos(x)$ meets the line $y = x$.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def g(x):
    return np.cos(x)

n = 5
x0 = 3 * np.pi

xs = np.linspace(-5 / 2 * np.pi, 7 / 2 * np.pi, 1000)

fig, ax = plt.subplots(figsize=(10, 4))
ax.axhline(0, color="gray", linewidth=0.5)
ax.axvline(0, color="gray", linewidth=0.5)
ax.plot(xs, g(xs), label="$y = \\cos(x)$")
ax.plot(xs, xs, label="$y = x$")

x = x0
for k in range(n + 1):
    ax.plot(x, g(x), "ko", clip_on=False)
    ax.annotate(f"$x_{{{k}}}$", (x, g(x)), textcoords="offset points", xytext=(5, 5))
    x = g(x)

ax.set_xlim(xs[0], xs[-1])
ax.set_ylim(-1, 1)
ticks = np.arange(-2, 4) * np.pi
ax.set_xticks(ticks, ["$-2\\pi$", "$-\\pi$", "$0$", "$\\pi$", "$2\\pi$", "$3\\pi$"])
ax.legend()
plt.show()
```

# Newton and Secant Methods

We implement Newton's method and the secant method following the pseudocode in the lecture notes, and compare them with regula falsi.

## Newton's method

```python
def newton(f, df, x, ftol, xtol, maxit):
    print(f"{'n':<3} |  {'x_n':<22} |  {'f(x_n)':<10} |  |x_n - x_(n-1)|")
    print("-" * 64)
    fx = f(x)
    print(f"{0:<3d} | {x:< 23.16e} | {fx:< 11.4e} |")
    for n in range(1, maxit + 1):
        delta = fx / df(x)
        x = x - delta
        fx = f(x)
        print(f"{n:<3d} | {x:< 23.16e} | {fx:< 11.4e} | {abs(delta):< .4e}")
        if abs(fx) <= ftol or abs(delta) <= xtol:
            return x
    print("Newton's method did not converge in maximum number of iterations")
    return x
```

## Secant method

The iterates are numbered starting from the two initial guesses $x_0$ and $x_1$. Each step needs only one new evaluation of $f$, since $f(x_{n-1})$ is kept from the previous step.

```python
def secant(f, x0, x1, ftol, xtol, maxit):
    print(f"{'n':<3} |  {'x_n':<22} |  {'f(x_n)':<10} |  |x_n - x_(n-1)|")
    print("-" * 64)
    f0 = f(x0)
    f1 = f(x1)
    print(f"{0:<3d} | {x0:< 23.16e} | {f0:< 11.4e} |")
    print(f"{1:<3d} | {x1:< 23.16e} | {f1:< 11.4e} | {abs(x1 - x0):< .4e}")
    for n in range(2, maxit + 2):
        delta = (x1 - x0) * f1 / (f1 - f0)
        x0, f0 = x1, f1
        x1 = x1 - delta
        f1 = f(x1)
        print(f"{n:<3d} | {x1:< 23.16e} | {f1:< 11.4e} | {abs(delta):< .4e}")
        if abs(f1) <= ftol or abs(delta) <= xtol:
            return x1
    print("Secant method did not converge in maximum number of iterations")
    return x1
```

## Regula falsi

The implementation from the root bracketing notebook.

```python
def regula_falsi(f, a, b, ftol, xtol, maxit):
    print(f"{'n':<3} |  {'x_n':<22} |  {'f(x_n)':<10} |  bound on |x_n - r|")
    print("-" * 64)
    fa = f(a)
    fb = f(b)
    x = (a * fb - b * fa) / (fb - fa)
    fx = f(x)
    print(f"{0:<3d} | {x:< 23.16e} | {fx:< 11.4e} | {max(x - a, b - x):< .4e}")
    for n in range(1, maxit + 1):
        if fa * fx < 0:
            b = x
            fb = fx
        else:
            a = x
            fa = fx
        x = (a * fb - b * fa) / (fb - fa)
        fx = f(x)
        print(f"{n:<3d} | {x:< 23.16e} | {fx:< 11.4e} | {max(x - a, b - x):< .4e}")
        if abs(fx) <= ftol or max(x - a, b - x) <= xtol:
            return x
    print("Regula falsi did not converge in maximum number of iterations")
    return x
```

## Comparison: secant method and regula falsi

Example 2.11 (Table 2.3): $f(x) = x^2 - 2$, whose positive root is $r = \sqrt{2}$. For regula falsi we use the interval $[1, 2]$, and for the secant method the initial guesses $x_0 = 1$, $x_1 = 2$.

```python
def f(x):
    return x**2 - 2

def df(x):
    return 2 * x
```

```python
x = regula_falsi(f, 1.0, 2.0, ftol=1e-14, xtol=1e-14, maxit=100)
```

```python
x = secant(f, 1.0, 2.0, ftol=1e-14, xtol=1e-14, maxit=100)
```

For comparison, Newton's method with initial guess $x_0 = 1$.

```python
x = newton(f, df, 1.0, ftol=1e-14, xtol=1e-14, maxit=100)
```

## Comparison: Newton's method and the secant method

Example 2.14 (Table 2.4): $f(x) = x^3 - 2x + 2$, whose only real root is
$$
   r = -\left( \left(1 + \frac{\sqrt{57}}{9}\right)^{1/3} + \left(1 - \frac{\sqrt{57}}{9}\right)^{1/3} \right) \approx -1.769292354238631.
$$
For Newton's method we use the initial guess $x_0 = -2$, and for the secant method we use the first two Newton iterates, $x_0 = -2$ and $x_1 = -1.8$.

```python
def g(x):
    return x**3 - 2 * x + 2

def dg(x):
    return 3 * x**2 - 2
```

```python
x = newton(g, dg, -2.0, ftol=1e-14, xtol=1e-14, maxit=100)
```

```python
x = secant(g, -2.0, -1.8, ftol=1e-14, xtol=1e-14, maxit=100)
```

## Divergence of Newton's method

Example 2.15 (Table 2.5): $f(x) = \arctan(x)$, whose only root is $r = 0$. Starting from $x_0 = 1$, Newton's method converges, but starting from $x_0 = \sqrt{3}$, the iterates alternate in sign and grow rapidly in magnitude. We limit the number of iterations to 10, since the iterates soon overflow.

```python
import math

def h(x):
    return math.atan(x)

def dh(x):
    return 1 / (1 + x**2)
```

```python
x = newton(h, dh, 1.0, ftol=1e-14, xtol=1e-14, maxit=10)
```

```python
x = newton(h, dh, math.sqrt(3), ftol=1e-14, xtol=1e-14, maxit=10)
```

## Interactive Newton's method

Each click of the button takes one step of Newton's method. The plot shows $f$, the tangent line at the current approximation $x_n$, and its root, which is the next approximation $x_{n+1}$.

The plot is interactive only when the notebook is running (in JupyterLite or Colab), not in the static HTML version.

```python
# In JupyterLite, ipywidgets must be installed in the browser before it is imported
import sys

if sys.platform == "emscripten":
    import piplite
    await piplite.install("ipywidgets")
```

```python
import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import display

def interactive_newton(f, df, x, ftol, xtol, maxit):
    n = 0
    fx = f(x)
    iterates = [x]
    plot = widgets.Output()
    status = widgets.HTML()
    button = widgets.Button(description="Take one step")

    def stop(message):
        button.disabled = True
        status.value += f"<br><b>{message}</b>"

    def draw():
        # The tangent line at x_n and its root x_{n+1} (if they can be computed)
        try:
            dfx = df(x)
            x_next = x - fx / dfx
        except ArithmeticError:
            dfx = x_next = None
        # Zoom in on the current step, from x_n to x_{n+1}
        width = abs(x_next - x) if x_next is not None else 0.0
        width = max(width, 1e-6 * (1 + abs(x)))
        lo = min(x, x_next if x_next is not None else x) - 0.5 * width
        hi = max(x, x_next if x_next is not None else x) + 0.5 * width
        t = np.linspace(lo, hi, 400)
        with plot:
            plot.clear_output(wait=True)
            fig, ax = plt.subplots(figsize=(7, 4))
            ax.axhline(0, color="black", linewidth=0.8)
            ax.plot(t, [f(ti) for ti in t], label="$f(x)$")
            previous = [xi for xi in iterates[:-1] if lo <= xi <= hi]
            ax.plot(previous, [0] * len(previous), "o", color="gray", alpha=0.5, label="previous iterates")
            if x_next is not None:
                ax.plot(t, fx + dfx * (t - x), "--", label="tangent line at $x_n$")
                ax.plot([x_next], [0], "o", color="C3", label="root of tangent line, $x_{n+1}$")
            ax.plot([x], [fx], "o", color="C1", label="$(x_n, f(x_n))$")
            ax.set_xlim(lo, hi)
            ax.legend(loc="best")
            plt.show()
        status.value = (f"Iteration <i>n</i> = {n}<br>"
                        f"Approximation <i>x<sub>n</sub></i> = {x:.16e}<br>"
                        f"Residual <i>f</i>(<i>x<sub>n</sub></i>) = {fx:.4e}")
        return x_next

    def step(_):
        nonlocal n, x, fx
        x_next = x - fx / df(x)
        delta = x - x_next
        x = x_next
        fx = f(x)
        n += 1
        iterates.append(x)
        x_next = draw()
        if abs(fx) <= ftol or abs(delta) <= xtol:
            stop("Converged.")
        elif n >= maxit:
            stop("Newton's method did not converge in maximum number of iterations.")
        elif x_next is None:
            stop("The tangent line at x<sub>n</sub> cannot be computed (horizontal, or overflow).")

    button.on_click(step)
    display(widgets.VBox([plot, status, button]))
    if draw() is None:
        stop("The tangent line at x<sub>n</sub> cannot be computed (horizontal, or overflow).")
```

Example 2.14, starting from $x_0 = -2$.

```python
interactive_newton(g, dg, -2.0, ftol=1e-14, xtol=1e-14, maxit=100)
```

Example 2.15, starting from $x_0 = \sqrt{3}$, where Newton's method diverges.

```python
interactive_newton(h, dh, math.sqrt(3), ftol=1e-14, xtol=1e-14, maxit=10)
```

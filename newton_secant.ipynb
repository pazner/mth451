# Newton and Secant Methods

We implement Newton's method and the secant method following the pseudocode in the lecture notes, and compare them with regula falsi.

## Newton's method

```python
def newton(f, df, x, ftol, xtol, maxit):
    print(f"{'n':<3} |  {'x_n':<22} |  {'f(x_n)':<10} |  |x_n - x_(n-1)|")
    print("-" * 64)
    fx = f(x)
    print(f"{0:<3d} | {x:< 23.16e} | {fx:< 11.4e} |")
    for n in range(1, maxit + 1):
        delta = fx / df(x)
        x = x - delta
        fx = f(x)
        print(f"{n:<3d} | {x:< 23.16e} | {fx:< 11.4e} | {abs(delta):< .4e}")
        if abs(fx) <= ftol or abs(delta) <= xtol:
            return x
    print("Newton's method did not converge in maximum number of iterations")
    return x
```

## Secant method

The iterates are numbered starting from the two initial guesses $x_0$ and $x_1$. Each step needs only one new evaluation of $f$, since $f(x_{n-1})$ is kept from the previous step.

```python
def secant(f, x0, x1, ftol, xtol, maxit):
    print(f"{'n':<3} |  {'x_n':<22} |  {'f(x_n)':<10} |  |x_n - x_(n-1)|")
    print("-" * 64)
    f0 = f(x0)
    f1 = f(x1)
    print(f"{0:<3d} | {x0:< 23.16e} | {f0:< 11.4e} |")
    print(f"{1:<3d} | {x1:< 23.16e} | {f1:< 11.4e} | {abs(x1 - x0):< .4e}")
    for n in range(2, maxit + 2):
        delta = (x1 - x0) * f1 / (f1 - f0)
        x0, f0 = x1, f1
        x1 = x1 - delta
        f1 = f(x1)
        print(f"{n:<3d} | {x1:< 23.16e} | {f1:< 11.4e} | {abs(delta):< .4e}")
        if abs(f1) <= ftol or abs(delta) <= xtol:
            return x1
    print("Secant method did not converge in maximum number of iterations")
    return x1
```

## Regula falsi

The implementation from the root bracketing notebook.

```python
def regula_falsi(f, a, b, ftol, xtol, maxit):
    print(f"{'n':<3} |  {'x_n':<22} |  {'f(x_n)':<10} |  bound on |x_n - r|")
    print("-" * 64)
    fa = f(a)
    fb = f(b)
    x = (a * fb - b * fa) / (fb - fa)
    fx = f(x)
    print(f"{0:<3d} | {x:< 23.16e} | {fx:< 11.4e} | {max(x - a, b - x):< .4e}")
    for n in range(1, maxit + 1):
        if fa * fx < 0:
            b = x
            fb = fx
        else:
            a = x
            fa = fx
        x = (a * fb - b * fa) / (fb - fa)
        fx = f(x)
        print(f"{n:<3d} | {x:< 23.16e} | {fx:< 11.4e} | {max(x - a, b - x):< .4e}")
        if abs(fx) <= ftol or max(x - a, b - x) <= xtol:
            return x
    print("Regula falsi did not converge in maximum number of iterations")
    return x
```

## Comparison: secant method and regula falsi

Example 2.11 (Table 2.3): $f(x) = x^2 - 2$, whose positive root is $r = \sqrt{2}$. For regula falsi we use the interval $[1, 2]$, and for the secant method the initial guesses $x_0 = 1$, $x_1 = 2$.

```python
def f(x):
    return x**2 - 2

def df(x):
    return 2 * x
```

```python
x = regula_falsi(f, 1.0, 2.0, ftol=1e-14, xtol=1e-14, maxit=100)
```

```python
x = secant(f, 1.0, 2.0, ftol=1e-14, xtol=1e-14, maxit=100)
```

For comparison, Newton's method with initial guess $x_0 = 1$.

```python
x = newton(f, df, 1.0, ftol=1e-14, xtol=1e-14, maxit=100)
```
